```python
import pandas as pd

df = pd.read_csv('disponibilità/A&T_Disp_2WheelsPoliTO.csv', sep=';')
df
```

```python
import config
from models import Persona
```

```python
df['MeMatt'] = df['MeMatt'].rename('Mercoledì Mattino')
df['MePom'] = df['MePom'].rename('Mercoledì Pomeriggio')
df['GioMatt'] = df['GioMatt'].rename('Giovedì Mattino')
df['GioPomm'] = df['GioPomm'].rename('Giovedì Pomeriggio')
df['VeMatt'] = df['VeMatt'].rename('Venerdì Mattino')
df['VePom'] = df['VePom'].rename('Venerdì Pomeriggio')

string_to_bool = {"SI": True, "NO": False}
df['NEW 2026'] = df['NEW 2026'].map(string_to_bool)

persone = []
for idx, row in df.iterrows():
    persona = Persona(
        matricola=str(row[config.MATRICOLA]),
        nome=row[config.NOME],
        cognome=row[config.COGNOME],
        is_esperto=(row[config.ESPERTO] == False)  # 0 = esperto, 1 = nuovo
    )
    persone.append(persona)

persone
```
